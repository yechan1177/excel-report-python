## 2. 이번 주 폴더에 어떤 파일이 있나
`WEEK`는 보고할 주의 월요일 날짜다. 지점 파일은 `data/raw/<월요일 날짜>/` 폴더에 모아 둔다.

```python
from pathlib import Path

DATA = Path("../data")
WEEK = "2026-11-02"

week_dir = DATA / "raw" / WEEK
for f in sorted(week_dir.glob("*.xlsx")):
    print(f.name)
```

## 3. 파일 하나 읽기
강남점 파일은 첫 줄이 열 이름이고 그 아래가 바로 데이터라서 그냥 읽힌다.

```python
df = pd.read_excel(week_dir / "강남점_주간판매.xlsx")
df.head()
```

열마다 무엇으로 읽혔는지 본다. 날짜는 `datetime64`, 숫자는 `int64`로 읽혀야 계산이 된다.

```python
df.dtypes
```

## 4. 엑셀에서 하던 것 두 가지
합계(SUM)와 품목별 합계(피벗테이블)를 한 줄씩으로 해 본다.

```python
print("판매 건수:", len(df))
print("금액 합계:", f"{df['금액'].sum():,}원")
```

```python
df.groupby("품목명")["금액"].sum().sort_values(ascending=False).head(5)
```
